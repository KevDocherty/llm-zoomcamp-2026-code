# Choosing an Agent Framework

Video: [Watch this lesson](https://www.youtube.com/watch?v=4yiCbKX9RhI&list=PL3MmuxUbc_hLZFNgSad56pDBKK8KO0XIv)

The core ideas from Part 2 transfer across frameworks: define tools, send messages to a model, execute requested tool calls, return their results, and continue until the model finishes.

The same model–tool feedback cycle appears across frameworks, but their higher-level features differ. Important differences include state management, validation, concurrency, observability, human approval, durable execution, and deployment support. Understanding the underlying loop makes those abstractions easier to evaluate.

Here are several frameworks worth exploring.


## OpenAI Agents SDK

The [OpenAI Agents SDK](https://openai.github.io/openai-agents-python/) is OpenAI's official Python framework for agent applications. It uses the Responses API by default and adds a managed agent loop, function tools, argument validation, sessions, guardrails, tracing, and handoffs between agents.

```bash
uv add openai-agents
```

It is a strong starting point when you already use OpenAI and want a lightweight framework with official support.


## PydanticAI

[PydanticAI](https://pydantic.dev/docs/ai/) is a Python agent framework with a strong emphasis on type-safe dependencies, validated structured outputs, and support for multiple model providers. Tools can be ordinary Python functions whose annotations and docstrings describe their inputs.

```bash
uv add pydantic-ai
```

Many frameworks validate tool arguments, but PydanticAI places particular emphasis on end-to-end typing and structured-output validation. It is a good choice when those features and provider flexibility are priorities.


## LangChain and LangGraph

[LangChain](https://docs.langchain.com/) provides model, tool, retrieval, and integration abstractions. [LangGraph](https://docs.langchain.com/oss/python/langgraph/overview) adds explicit graph-based orchestration for stateful workflows whose steps, branches, or interruptions need to be controlled.

```bash
uv add langchain langgraph
```

They are useful when you need a large integration ecosystem or want to model a workflow explicitly as a graph.


## Google ADK

Google's [Agent Development Kit](https://adk.dev/) provides familiar building blocks such as agents, tools, instructions, sessions, and multi-agent workflows. It integrates naturally with Gemini and Google Cloud services.

```bash
uv add google-adk
```

It is a natural option when your application is built around Gemini or Google Cloud.


## A quick selection guide

| Need | Useful starting point |
| --- | --- |
| An OpenAI-first application with a managed loop | OpenAI Agents SDK |
| Strong Python typing and provider flexibility | PydanticAI |
| Explicit, stateful graph workflows | LangGraph |
| Gemini and Google Cloud integration | Google ADK |
| Lightweight learning and experimentation | ToyAIKit or smolagents |

This table is a starting point, not a ranking. Before choosing, build a small version of your own workflow and compare developer experience, tracing, testing, deployment, and provider support.


## Other options

Other frameworks worth knowing include:

- **CrewAI** — role-based multi-agent orchestration
- **AutoGen** — multi-agent applications and conversations
- **Semantic Kernel** — Microsoft's SDK for Python, C#, and Java
- **smolagents** — a lightweight agent library from Hugging Face

Provider-native tool-use APIs are another option when you want to own the loop yourself. Examples include OpenAI's Responses API and Anthropic's tool-use API. These are APIs rather than agent frameworks, so your application remains responsible for tool dispatch, message history, stopping conditions, and error handling.


## Avoiding agents when you can

You do not always need an agent. Adding one introduces real costs:

- More API calls, because the loop may make several tool calls before finishing
- Higher latency from repeated model round-trips
- Higher spend as conversation history and tool results accumulate
- More operational work, including cost, iteration, and failure monitoring
- Less predictable behavior, because the model chooses the next action

Start by asking whether the workflow can be expressed more simply:

- Use plain RAG when every request follows one retrieval step and one answer step.
- Use ordinary code for deterministic parsing, transformation, or templating.
- Use a single LLM call when no external actions or iterative decisions are needed.

An agent becomes useful when the application must choose dynamically among tools, revise its approach after seeing intermediate results, or handle tasks whose steps cannot be fixed reliably in advance.

When an agent is justified, add practical safeguards: validate tool arguments, restrict permissions, cap iterations and cost, handle failures, record traces, and test behavior against representative examples.


## Key takeaway

Framework choice matters less than understanding the workflow you need. Start with the simplest design that solves the problem, then adopt a framework when its state management, validation, observability, or orchestration features remove meaningful complexity.
